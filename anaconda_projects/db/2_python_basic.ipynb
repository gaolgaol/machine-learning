# 2.3 자료형

## 2.3.1 자료형의 종류

| 자료형 | 사용 예 | 자료형의 의미 |
|:---:|:---|:---|
| int | `a = 10` | 정수형. 소수점이 없는 정수를 나타냄 |
| float | `a = 3.14` | 실수형. 소수점이 있는 숫자를 나타냄 |
| str | `a = "Hello"` | 문자열형. 문자나 문자열을 나타냄 |
| bool | `a = True` | 논리형. 참(True) 또는 거짓(False)을 나타냄 |
| list | `a = [1, 2, 3]` | 리스트형. 여러 데이터를 순서대로 저장하며 수정 가능 |
| tuple | `a = (1, 2, 3)` | 튜플형. 여러 데이터를 순서대로 저장하며 수정 불가능 |
| ndarray | `a = np.array([1, 2, 3])` | NumPy 배열형. 다차원 배열을 표현하고 수치 연산에 활용 |
